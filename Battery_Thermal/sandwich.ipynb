```python
area = 6.25e-4 #m^2
q_in = 80 #W

# copper
t_cu = .1e-3 #m
k_cu = 400 #w/mK
R_cu = t_cu / (k_cu * area)

# silicone
t_si = .9e-3 #m
k_si = 6 #w/mK
R_si = t_si / (k_si * area)

# aluminum
t_al = 2.5e-3 #m
k_al = 138 #w/mK
R_al = t_al / (k_al * area)


R_total = R_cu + R_si + R_al
print(f'R_total (K/W): ', R_total)

delta_T = q_in * R_total
print(f'delta_T (C): ', delta_T)
```
